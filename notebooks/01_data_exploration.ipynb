## Pull Team Stats


```python
from nba_api.stats.endpoints import leaguedashteamstats

team_stats = leaguedashteamstats.LeagueDashTeamStats(season='2025-26').get_data_frames()[0]
team_stats
```

## Save Raw Data


```python
team_stats.to_csv("../data/raw/team_stats_2025_26.csv", index=False)
```

## Check & Clean
